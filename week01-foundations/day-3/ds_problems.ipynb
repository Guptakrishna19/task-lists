1) Remove duplicates from a list
   time - O(n) | space - O(n)

```python
def remove_duplicates(items):
    return list(set(items))

nums = [1, 2, 2, 3, 4, 4, 4]
print(remove_duplicates(nums))
```

2) Dedupe while preserving order
time - O(n) | space - O(n)

```python
def dedupe_ordered(items):
    seen = set()
    result = []
    for x in items:
        if x not in seen: 
            result.append(x)
            seen.add(x)
    return result

print(dedupe_ordered([3, 1, 2, 1, 3]))
```

3) Count word frequencies
time - O(n) | space - O(k)

```python
from collections import Counter

def word_count(words):
    freq = {}
    for w in words:
        freq[w] = freq.get(w, 0) + 1
    return freq

def word_count_v2(words):
    return dict(Counter(words))

words = ["apple", "banana", "apple", "cherry", "banana", "apple"]
print(word_count(words))
```

4) Top-k most frequent elements
time - O(n) | space - O(k)


```python
from collections import Counter

def top_k_frequent(nums, k):
    freq = Counter(nums)
    return [val for val, _ in freq.most_common(k)]

nums = [1, 1, 1, 2, 2, 3]
print(top_k_frequent(nums, 2))
```

5) Two-sum
time - O(n) | space - O(n)

```python
def two_sum(nums, target):
    seen = {} 
    for i, x in enumerate(nums):
        complement = target - x
        if complement in seen:
            return [seen[complement], i]
        seen[x] = i
    return [] 

print(two_sum([2, 7, 11, 15], 9))
```

6) Two-sum: return values not indices
time - O(n) | space - O(n)

```python
def two_sum_values(nums, target):
    seen = set()
    for x in nums:
        complement = target - x
        if complement in seen:
            return (complement, x)
        seen.add(x)
    return None

print(two_sum_values([4, 6, 1, 8], 10))
```

7) Group strings by first letter
time - O(n) | space - O(n)

```python
from collections import defaultdict

def group_by_letter(words):
    groups = defaultdict(list)
    for w in words:
        groups[w[0].lower()].append(w)
    return dict(groups)

words = ["apple", "avocado", "banana", "blueberry", "cherry"]
print(group_by_letter(words))
```

8) Group anagrams together
time - O(n · m logm) | space - O(n)

```python
from collections import defaultdict

def group_anagrams(words):
    groups = defaultdict(list)
    for w in words:
        key = "".join(sorted(w)) 
        groups[key].append(w)
    return list(groups.values())

words = ["eat", "tea", "tan", "ate", "nat", "bat"]
print(group_anagrams(words))
```

9) Sort list of dicts by a key
time - O(n log n)    | space - O(n)

```python
def sort_by_key(records, field, reverse=False):
    return sorted(records, key=lambda r: r[field], reverse=reverse)

students = [
    {"name": "Anika", "grade": 88},
    {"name": "Rohan", "grade": 95},
    {"name": "Meera", "grade": 72},
]
print(sort_by_key(students, "grade"))

print(sort_by_key(students, "grade", reverse=True))
```

10) Sort by multiple keys
time - O(n log n)    | space - O(n)

```python
def sort_multi_key(records):
    return sorted(records, key=lambda r: (r["dept"], -r["salary"]))

employees = [
    {"name": "Alice", "dept": "Eng", "salary": 90000},
    {"name": "Bob",   "dept": "Eng", "salary": 85000},
    {"name": "Carol", "dept": "HR",  "salary": 70000},
    {"name": "Dave",  "dept": "HR",  "salary": 75000},
]
for e in sort_multi_key(employees):
    print(e["name"], e["dept"], e["salary"])
```
