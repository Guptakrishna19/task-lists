### Exercise 6: Indexing

```python
# Access element at index 3
arr = np.arange(10)
element = arr[3]
print(f"Array: {arr}")
print(f"Element at index 3: {element}")
```

### Exercise 7: Slicing

```python
# Slice elements from index 2 to 6 (inclusive of 2, exclusive of 7)
arr = np.arange(10)
sliced = arr[2:7]
print(f"Array: {arr}")
print(f"Slice [2:7]: {sliced}")
```

### Exercise 8: Boolean Masking

```python
# Filter elements greater than 10
arr = np.arange(20)
mask = arr > 10
filtered = arr[mask]
print(f"Array: {arr}")
print(f"Elements > 10: {filtered}")
```

### Exercise 9: Complex Filtering (Conditional)

```python
# Filter even numbers
arr = np.arange(20)
even_mask = (arr % 2) == 0
even_numbers = arr[even_mask]
print(f"Array: {arr}")
print(f"Even numbers: {even_numbers}")
```

### Exercise 10: Aggregations

```python
# Compute sum, mean, max, min
arr = np.arange(1, 11)
print(f"Array: {arr}")
print(f"Sum: {arr.sum()}")
print(f"Mean: {arr.mean()}")
print(f"Median: {np.median(arr)}")
print(f"Std Dev: {arr.std()}")
print(f"Max: {arr.max()}")
print(f"Min: {arr.min()}")
```

### Exercise 11: Axis Operations

```python
# Sum along different axes
mat = np.arange(12).reshape(3, 4)
print(f"Matrix (3x4):\n{mat}\n")
print(f"Sum along axis 0 (columns): {mat.sum(axis=0)}")
print(f"Sum along axis 1 (rows): {mat.sum(axis=1)}")
print(f"\nMean along axis 0: {mat.mean(axis=0)}")
print(f"Mean along axis 1: {mat.mean(axis=1)}")
```

### Exercise 12: Broadcasting

```python
# Broadcasting: add scalar and array to matrix
mat = np.array([[1, 2, 3], [4, 5, 6]])
print(f"Original matrix:\n{mat}\n")

# Add scalar
result1 = mat + 10
print(f"Matrix + 10:\n{result1}\n")

# Add vector (broadcasts)
vec = np.array([1, 2, 3])
result2 = mat + vec
print(f"Matrix + [1,2,3] (broadcast):\n{result2}")
```

## Benchmarking: Python Loops vs NumPy Vectorization

### Setup: Create test data

```python
# Create large arrays for benchmarking
n = 1_000_000
np_array = np.arange(n)
py_list = list(range(n))

print(f"Array size: {n:,} elements")
```

### Benchmark 1: Summing All Elements

```python
# Python loop sum
def py_sum(lst):
    total = 0
    for x in lst:
        total += x
    return total

# Benchmark
py_time = timeit(lambda: py_sum(py_list), number=10)
np_time = timeit(lambda: np_array.sum(), number=10)

speedup = py_time / np_time
print(f"Python loop sum: {py_time:.6f}s")
print(f"NumPy sum:       {np_time:.6f}s")
print(f"Speedup:         {speedup:.2f}x")
```

### Benchmark 2: Element-wise Multiplication

```python
# Python loop multiplication
def py_multiply_scalar(lst, scalar):
    return [x * scalar for x in lst]

# Benchmark
py_time = timeit(lambda: py_multiply_scalar(py_list, 2.5), number=10)
np_time = timeit(lambda: np_array * 2.5, number=10)

speedup = py_time / np_time
print(f"Python loop multiply: {py_time:.6f}s")
print(f"NumPy multiply:       {np_time:.6f}s")
print(f"Speedup:              {speedup:.2f}x")
```

### Benchmark 3: Element-wise Addition of Two Arrays

```python
# Create second arrays
np_array2 = np.arange(n, 2*n)
py_list2 = list(range(n, 2*n))

# Python loop addition
def py_add_arrays(lst1, lst2):
    return [a + b for a, b in zip(lst1, lst2)]

# Benchmark
py_time = timeit(lambda: py_add_arrays(py_list, py_list2), number=10)
np_time = timeit(lambda: np_array + np_array2, number=10)

speedup = py_time / np_time
print(f"Python loop add:  {py_time:.6f}s")
print(f"NumPy add:        {np_time:.6f}s")
print(f"Speedup:          {speedup:.2f}x")
```

### Benchmark 4: Filtering with Condition

```python
# Python loop filtering
def py_filter_even(lst):
    return [x for x in lst if x % 2 == 0]

# Benchmark
py_time = timeit(lambda: py_filter_even(py_list), number=10)
np_time = timeit(lambda: np_array[np_array % 2 == 0], number=10)

speedup = py_time / np_time
print(f"Python loop filter: {py_time:.6f}s")
print(f"NumPy filter:       {np_time:.6f}s")
print(f"Speedup:            {speedup:.2f}x")
```

## Benchmark Summary & Analysis

### Results

**Key Findings:**

1. **Sum operation**: NumPy is ~50-100x faster than Python loops
   - Python: Requires explicit iteration and accumulation
   - NumPy: Uses vectorized C operations under the hood

2. **Element-wise multiplication**: NumPy is ~30-60x faster
   - Python: List comprehension still requires iteration
   - NumPy: Operates on entire array at once in compiled C

3. **Array addition**: NumPy is ~40-80x faster
   - Similar pattern: zip iteration vs vectorized operation

4. **Filtering**: NumPy is ~20-40x faster
   - Python: List comprehension with condition check per element
   - NumPy: Boolean mask applied efficiently

### Why NumPy is Fast

- **Vectorization**: Operations work on entire arrays without explicit Python loops
- **Memory efficiency**: Fixed-size, homogeneous arrays vs dynamic Python lists
- **C implementation**: Core operations in compiled C, not interpreted Python
- **Cache locality**: Contiguous memory layout improves CPU cache efficiency
- **SIMD**: Modern CPUs process multiple elements in parallel

### When to Use NumPy

✅ Large arrays (1000+ elements)
✅ Numerical computations (math, science, data analysis)
✅ Need performance-critical code
✅ Processing images, audio, or matrices

❌ Small datasets or simple operations
❌ Complex object manipulation
❌ When Python's simplicity is more important than speed
