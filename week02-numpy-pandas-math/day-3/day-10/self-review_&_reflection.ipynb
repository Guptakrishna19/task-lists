## 4. Self-Review & Reflection

**Key things I learned today:**
- `loc` selects by **label**; `iloc` selects by **integer position**
- Boolean filters use True/False masks — always wrap conditions in `()`
- `groupby().agg()` is the workhorse for any "summarize by group" question
- `merge` joins two DataFrames on a shared key — just like SQL JOIN
- `sort_values` orders rows; `ascending=False` = largest first
- Missing values (`NaN`) need to be handled before aggregating (`dropna` or `fillna`)

**What was difficult:**
- The difference between `loc` and `iloc`
- Chaining `&`/`|` conditions — forgetting the `()`
- `reset_index()` after `groupby` — why is it needed?
- `pct_change()` for growth calculations

**What I improved / fixed:**
- Fixed Boolean condition syntax: Initially forgot parentheses around conditions like (orders['category'] == 'Electronics') & (orders['revenue'] > 1000) and got errors — now I remember to wrap each condition in () when using & or |.
- Understood reset_index(): After groupby().agg(), the result has a grouped index (like 'region' as row labels). Using .reset_index() turns that index back into a normal column so I can keep working with the data (e.g., sort or merge it).
- Handled NaN values confidently: Saw 5 missing discounts in the data and learned to use .dropna(subset=['discount']) for calculations needing clean data (like average discount) vs. .fillna(0) when treating missing as "no discount".
- Chained operations smoothly: Gone from writing 5 separate lines to fluently chaining: orders.groupby('region')['revenue'].sum().reset_index().sort_values(..., ascending=False).
- Clicked loc vs iloc: loc uses labels (column names like 'product'), iloc uses positions (0, 1, 2...). No more guessing — I check if I'm selecting by name or by count.
