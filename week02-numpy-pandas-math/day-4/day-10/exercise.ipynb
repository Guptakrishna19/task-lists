## Exercise 1: Missing Data Handling

Practice with `isna()`, `fillna()`, and `dropna()` methods.

```python
import pandas as pd
import numpy as np

# Create sample data with missing values
df = pd.DataFrame({
    'A': [1, 2, np.nan, 4, 5],
    'B': [np.nan, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan]
})
print("Original DataFrame:")
print(df)
print()
print("Missing values:")
print(df.isna().sum())
print()
# Your turn: Practice fillna() and dropna()
# Example: df_filled = df.fillna(0)
# Example: df_dropped = df.dropna()
```

## Exercise 2: Apply and Map Functions

Practice using `apply()` and `map()` for data transformation.

```python
# Create sample data
df = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie'],
    'score': [85, 92, 78],
    'passed': [True, True, False]
})
print("Original DataFrame:")
print(df)
print()
# Your turn: Practice apply() and map()
# Example: df['score_double'] = df['score'].apply(lambda x: x * 2)
# Example: df['passed_str'] = df['passed'].map({True: 'Yes', False: 'No'})
```

## Exercise 3: DateTime Conversion and .dt Accessors

Practice converting strings to datetime and using .dt accessors.

```python
# Create sample data with date strings
df = pd.DataFrame({
    'date_str': ['2023-01-01', '2023-02-15', '2023-03-30'],
    'value': [100, 150, 200]
})
print("Original DataFrame:")
print(df)
print()
# Your turn: Practice pd.to_datetime() and .dt accessors
# Example: df['date'] = pd.to_datetime(df['date_str'])
# Example: df['year'] = df['date'].dt.year
# Example: df['month'] = df['date'].dt.month
```

## Exercise 4: Pivot Tables

Practice creating pivot tables to summarize data.

```python
# Create sample data
df = pd.DataFrame({
    'Region': ['North', 'South', 'East', 'West', 'North', 'South'],
    'Product': ['A', 'B', 'A', 'B', 'A', 'B'],
    'Sales': [100, 150, 200, 120, 180, 90]
})
print("Original DataFrame:")
print(df)
print()
# Your turn: Practice pivot_table()
# Example: pivot = df.pivot_table(values='Sales', index='Region', columns='Product', aggfunc='sum')
```

## Exercise 5: Method Chaining

Practice chaining multiple Pandas operations together.

```python
# Create sample data
df = pd.DataFrame({
    'age': [25, 30, 35, 40, 45, 50],
    'salary': [50000, 60000, 70000, 80000, 90000, 100000],
    'dept': ['HR', 'IT', 'Finance', 'HR', 'IT', 'Finance']
})
print("Original DataFrame:")
print(df)
print()
# Your turn: Practice method chaining
# Example: result = df.query('age > 30').groupby('dept')['salary'].mean()
```
