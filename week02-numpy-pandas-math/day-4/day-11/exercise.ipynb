## Exercise 1: Missing Data Handling

Practiced with `isna()`, `fillna()`, and `dropna()` methods.

```python
import pandas as pd
import numpy as np

# Create sample data with missing values
df = pd.DataFrame({
    'A': [1, 2, np.nan, 4, 5],
    'B': [np.nan, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan]
})
print("Original DataFrame:")
print(df)
print()

# Check for missing values - isna() 
print("Missing values:")
print(df.isna().sum())
print()

# Fill missing values with a specific value - fillna()
df_filled = df.fillna(0)
print("\nDataFrame after fillna(0):")
print(df_filled)
print()

# Drop rows with missing values - dropna()
df_dropped = df.dropna()
print("\nDataFrame after dropna():")
print(df_dropped)
```

## Exercise 2: Apply and Map Functions

Practiced using `apply()` and `map()` for data transformation.

```python
df = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie'],
    'score': [85, 92, 78],
    'passed': [True, True, False]
})
print("Original DataFrame:")
print(df)
print()

# Apply a function to a column - apply()
df['score_double'] = df['score'].apply(lambda x: x * 2)
print("\nAfter apply():")
print(df)
print()

# Map values in a column - map()
df['passed_str'] = df['passed'].map({
    True: 'Yes',
    False: 'No'
})
print("\nAfter map():")
print(df)
```

## Exercise 3: DateTime Conversion and .dt Accessors

Practiced converting strings to datetime and using .dt accessors.

```python
# Create sample data with date strings
df = pd.DataFrame({
    'date_str': ['2023-01-01', '2023-02-15', '2023-03-30'],
    'value': [100, 150, 200]
})
print("Original DataFrame:")
print(df)
print()

# Convert string dates to datetime objects - pd.to_datetime()
df['date'] = pd.to_datetime(df['date_str'])
print("\nAfter pd.to_datetime():")
print(df)
print()

# Extract year from datetime column - .dt.year
df['year'] = df['date'].dt.year
print("\nAfter extracting year:")
print(df)
print()

# Extract month from datetime column - .dt.month
df['month'] = df['date'].dt.month
print("\nAfter extracting month:")
print(df)
print()

# Extract day name from datetime column - .dt.day_name()
df['day_name'] = df['date'].dt.day_name()
print("\nAfter extracting day name:")
print(df)
```

## Exercise 4: Pivot Tables

Practiced creating pivot tables to summarize data.

```python
# Create sample data
df = pd.DataFrame({
    'Region': ['North', 'South', 'East', 'West', 'North', 'South'],
    'Product': ['A', 'B', 'A', 'B', 'A', 'B'],
    'Sales': [100, 150, 200, 120, 180, 90]
})
print("Original DataFrame:")
print(df)
print()

# Create a pivot table - pivot_table()
pivot = df.pivot_table(
    values='Sales',
    index='Region',
    columns='Product',
    aggfunc='sum'
)
print("\nPivot Table:")
print(pivot)
print()

# Fill missing values in the pivot table
pivot_filled = df.pivot_table(
    values='Sales',
    index='Region',
    columns='Product',
    aggfunc='sum',
    fill_value=0
)
print("\nPivot Table with fill_value=0:")
print(pivot_filled)
```

## Exercise 5: Method Chaining

Practice chaining multiple Pandas operations together.

```python
df = pd.DataFrame({
    'age': [25, 30, 35, 40, 45, 50],
    'salary': [50000, 60000, 70000, 80000, 90000, 100000],
    'dept': ['HR', 'IT', 'Finance', 'HR', 'IT', 'Finance']
})
print("Original DataFrame:")
print(df)
print()

# Method chaining: Filter rows and calculate average salary by department
result = (
    df
    .query('age > 30')
    .groupby('dept')['salary']
    .mean()
)
print("\nAverage salary by department (age > 30):")
print(result)
print()

# Method chaining: Add bonus column and sort by salary
result2 = (
    df
    .assign(bonus=lambda x: x['salary'] * 0.10)
    .sort_values('salary', ascending=False)
)
print("\nAfter assign() and sort_values():")
print(result2)
```
