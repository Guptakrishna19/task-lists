# Statistics Summary Task

## Objective: Statistical summary + 3 distribution plots + written interpretation


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
print('Libraries imported')
```

```python
# Generate data for analysis
np.random.seed(42)
# Normal distribution
normal = np.random.normal(loc=0, scale=1, size=500)
# Right-skewed distribution (exponential)
skewed_right = np.random.exponential(scale=1, size=500)
# Left-skewed distribution (beta)
skewed_left = np.random.beta(a=2, b=5, size=500)
# Create a DataFrame for easy summary
df = pd.DataFrame({
    'Normal': normal,
    'Skewed_Right': skewed_right,
    'Skewed_Left': skewed_left
})
print('Data generated. Shape:', df.shape)
```

```python
# Statistical summary
summary = df.describe().T
summary['skewness'] = df.skew()
summary['kurtosis'] = df.kurtosis()
summary['variance'] = df.var()
print('Statistical Summary:')
display(summary)
```

```python
# Correlation matrix (between the three variables)
corr_matrix = df.corr()
print('Correlation Matrix:')
display(corr_matrix)
```

```python
# Plot 1: Normal distribution
plt.figure(figsize=(10, 6))
plt.subplot(2, 2, 1)
sns.histplot(df['Normal'], kde=True, color='skyblue')
plt.title('Normal Distribution')
plt.xlabel('Value')
plt.ylabel('Frequency')
```

```python
# Plot 2: Right-skewed distribution
plt.subplot(2, 2, 2)
sns.histplot(df['Skewed_Right'], kde=True, color='salmon')
plt.title('Right-Skewed Distribution')
plt.xlabel('Value')
plt.ylabel('Frequency')
```

```python
# Plot 3: Left-skewed distribution
plt.subplot(2, 2, 3)
sns.histplot(df['Skewed_Left'], kde=True, color='lightgreen')
plt.title('Left-Skewed Distribution')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.tight_layout()
plt.show()
```

## Written Interpretation

**Skewness Interpretation:**
- The normal distribution has a skewness close to 0, indicating symmetry.
- The right-skewed distribution (exponential) has a positive skewness (typically >0), indicating a longer tail on the right side. This means there are a few unusually high values.
- The left-skewed distribution (beta) has a negative skewness (<0), indicating a longer tail on the left side, with a few unusually low values.

**Correlation Interpretation:**
- The correlation matrix shows the relationships between the three variables. Since they are generated independently, we expect correlations near 0.
- However, due to random sampling, there might be small non-zero correlations. In this case, the correlations should be close to zero, indicating no linear relationship.
- If we were to create variables with a known relationship (like study hours and exam scores), we would see stronger correlations (positive or negative).

In summary, skewness measures the asymmetry of the distribution, while correlation measures the linear relationship between two variables.
