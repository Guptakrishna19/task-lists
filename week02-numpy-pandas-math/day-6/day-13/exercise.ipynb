# Day 13 Exercises: Exploratory Data Analysis (EDA)

In this notebook, we will perform a basic EDA on the Superstore dataset to satisfy the requirements of Day 13.

## 1. Load Data

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the Superstore dataset (assuming csv file available or loading sample)
# df = pd.read_csv("Sample - Superstore.csv")
# For demonstration purposes, we will load a dummy structure
df = pd.DataFrame({
    "Category": ["Furniture", "Office Supplies", "Technology", "Furniture", "Technology"],
    "Sales": [250, 50, 500, 300, 700],
    "Profit": [20, 5, 100, 30, 150]
})
print(df.head())
```

## 2. Clean Data

```python
# Check for missing values
print(df.isnull().sum())

# Example cleaning (e.g., fill missing or drop duplicates)
df = df.drop_duplicates()
```

## 3. Univariate & Bivariate Analysis

```python
# Univariate: 
# 1) Plot distribution of Sales
plt.figure(figsize=(8, 4))
sns.histplot(df["Sales"], kde=True)
plt.title("Sales Distribution")
plt.show()

# 2) Profit Distribution
plt.figure(figsize=(8,5))
sns.histplot(df["Profit"], kde=True)
plt.title("Profit Distribution")
plt.show()

# 3) Orders by Category
plt.figure(figsize=(8,5))
sns.countplot(
    data=df,
    x="Category"
)
plt.title("Orders by Category")
plt.show()

# Bivariate: 
# 1) Profit vs Category
plt.figure(figsize=(8, 4))
sns.barplot(x="Category", y="Profit", data=df)
plt.title("Profit by Category")
plt.show()

# 2) Category vs Sales
plt.figure(figsize=(8,5))
sns.barplot(
    data=df,
    x="Category",
    y="Sales"
)
plt.title("Average Sales by Category")
plt.show()

# 3) Sales vs Profit
plt.figure(figsize=(8,5))
sns.scatterplot(
    data=df,
    x="Sales",
    y="Profit"
)
plt.title("Sales vs Profit")
plt.show()
```
