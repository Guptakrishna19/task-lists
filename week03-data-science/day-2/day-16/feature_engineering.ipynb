### 1. Binning
Binning helps capture non-linear relationships by dividing a continuous numerical feature into discrete intervals (bins). We show equal-width (`pd.cut`) and equal-frequency (`pd.qcut`) binning.

```python
import pandas as pd
import numpy as np

# Sample numerical data
df_bin = pd.DataFrame({'income': [20000, 35000, 55000, 80000, 120000, 150000, 250000]})

# Equal-width binning (pd.cut)
income_bins = [0, 50000, 100000, 200000, np.inf]
income_labels = ['Low', 'Medium', 'High', 'Very High']
df_bin['income_bracket'] = pd.cut(df_bin['income'], bins=income_bins, labels=income_labels)

# Equal-frequency binning (pd.qcut) - divides into 3 quantiles
df_bin['income_quantile'] = pd.qcut(df_bin['income'], q=3, labels=['Tier 3', 'Tier 2', 'Tier 1'])

print("=== Binning Demo ===")
print(df_bin)
```

### 2. Datetime Features
Extracting rich components from datetime fields such as hour, day of the week, and weekend flags.

```python
# Sample temporal data
df_date = pd.DataFrame({
    'timestamp': ['2026-06-18 08:30:00', '2026-06-20 17:45:00', '2026-06-21 22:15:00']
})

# Convert to datetime object
df_date['timestamp'] = pd.to_datetime(df_date['timestamp'])

# Extract features
df_date['hour'] = df_date['timestamp'].dt.hour
df_date['day_of_week'] = df_date['timestamp'].dt.dayofweek # Monday is 0, Sunday is 6
df_date['day_name'] = df_date['timestamp'].dt.day_name()
df_date['is_weekend'] = df_date['timestamp'].dt.dayofweek.isin([5, 6]).astype(int)

print("=== Datetime Features Demo ===")
print(df_date)
```

### 3. Ratios and Interaction Features
Creating interactions or ratios mathematically combines features to build meaningful business metrics.

```python
# Sample retail data
df_ratio = pd.DataFrame({
    'total_spent': [120.00, 350.00, 45.00, 600.00],
    'items_count': [4, 10, 1, 12]
})

# Ratio feature: average cost per item
df_ratio['avg_spent_per_item'] = df_ratio['total_spent'] / df_ratio['items_count']

print("=== Ratios & Interactions Demo ===")
print(df_ratio)
```

### 4. One-Hot vs. Ordinal Encoding
Properly converting text categories to numerical inputs based on order hierarchy.

```python
# Sample categorical data
df_cat = pd.DataFrame({
    'customer_tier': ['Silver', 'Gold', 'Bronze', 'Gold'], # Ordered
    'region': ['East', 'West', 'North', 'South']          # Unordered
})

# Ordinal encoding mapping (retaining logical order)
tier_map = {'Bronze': 0, 'Silver': 1, 'Gold': 2}
df_cat['customer_tier_encoded'] = df_cat['customer_tier'].map(tier_map)

# One-Hot encoding for nominal category (region)
df_encoded = pd.get_dummies(df_cat, columns=['region'], prefix='region', dtype=int)

print("=== Categorical Encoding Demo ===")
print(df_encoded)
```

### 5. Scaling
Scaling shifts numeric variables so that features with large ranges don't disproportionately influence distance-based models.

```python
from sklearn.preprocessing import StandardScaler, MinMaxScaler

# Sample numeric data with different scales
df_scale = pd.DataFrame({
    'age': [22, 58, 36],
    'yearly_salary': [45000, 140000, 85000]
})

# Standardization (Mean = 0, Std Dev = 1)
std_scaler = StandardScaler()
df_scale['salary_standardized'] = std_scaler.fit_transform(df_scale[['yearly_salary']])

# Normalization/Min-Max Scaling (Range: 0 to 1)
minmax_scaler = MinMaxScaler()
df_scale['age_normalized'] = minmax_scaler.fit_transform(df_scale[['age']])

print("=== Scaling Demo ===")
print(df_scale)
```
