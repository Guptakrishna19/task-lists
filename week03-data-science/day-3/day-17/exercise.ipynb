# Day 17 — Data Visualization Exercises

## Objective

Learn to:
- Choose the correct chart for a given problem.
- Create visualizations using Matplotlib and Seaborn.
- Identify misleading visualizations.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(42)
df = pd.DataFrame({
    "Month": range(1, 13),
    "Sales": np.random.randint(100, 500, 12),
    "Profit": np.random.randint(20, 150, 12),
    "Customers": np.random.randint(50, 300, 12)
})
```

## Exercise 1 — Line Chart

### Use When

Showing trends over time.

Examples:
- Monthly sales
- Website traffic
- Stock prices

```python
plt.figure(figsize=(7, 3.5))
plt.plot(df["Month"], df["Sales"], marker="o", color="#1f77b4")
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True, linestyle="--", alpha=0.6)
plt.show()
```

## Exercise 2 — Bar Chart

### Use When

Comparing categories.

Examples:
- Revenue by department
- Students by class

```python
plt.figure(figsize=(7, 3.5))
plt.bar(df["Month"], df["Profit"], color="#2ca02c")
plt.title("Monthly Profit")
plt.xlabel("Month")
plt.ylabel("Profit")
plt.show()
```

## Exercise 3 — Histogram

### Use When

Understanding distribution.

Questions:
- Is data normal?
- Is it skewed?

```python
plt.figure(figsize=(7, 3.5))
plt.hist(df["Sales"], bins=5, color="#d62728", edgecolor="white")
plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()
```

## Exercise 4 — Box Plot

### Use When

Finding:
- Median
- Spread
- Outliers

```python
plt.figure(figsize=(5, 3.5))
plt.boxplot(df["Sales"])
plt.title("Sales Box Plot")
plt.ylabel("Sales")
plt.show()
```

## Exercise 5 — Scatter Plot

### Use When

Studying relationships.

Examples:
- Height vs Weight
- Customers vs Sales

```python
plt.figure(figsize=(7, 3.5))
plt.scatter(df["Customers"], df["Sales"], color="#9467bd")
plt.xlabel("Customers")
plt.ylabel("Sales")
plt.title("Customers vs Sales")
plt.show()
```

## Exercise 6 — Heatmap

### Use When

Checking correlation.

```python
plt.figure(figsize=(6, 4))
sns.heatmap(df.corr(), annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()
```

## Exercise 7 — Pairplot

### Use When

Quick EDA.

```python
pair = sns.pairplot(df)
pair.fig.suptitle("Pairplot of All Variables", y=1.02)
plt.show()
```

## Exercise 8 — Annotated Chart

### Use When

Highlighting important points.

```python
plt.figure(figsize=(7, 3.5))
plt.plot(df["Month"], df["Sales"], marker="o")
max_idx = df["Sales"].idxmax()
plt.annotate(
    "Peak Sales",
    xy=(df["Month"][max_idx], df["Sales"][max_idx]),
    xytext=(df["Month"][max_idx] - 2, df["Sales"][max_idx] + 30),
    arrowprops=dict(facecolor='black', shrink=0.05, width=1, headwidth=6)
)
plt.title("Annotated Sales Chart")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

## Exercise 9 — Fix a Misleading Chart

### Problem
- Y-axis starts at 350.
- Differences appear much larger than they actually are.

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))

# Misleading (Truncated Axis)
ax1.bar(df["Month"], df["Sales"], color="red")
ax1.set_ylim(350, 500)
ax1.set_title("Misleading (Truncated Axis)")

# Correct (Full Axis)
ax2.bar(df["Month"], df["Sales"], color="green")
ax2.set_ylim(0, df["Sales"].max() + 50)
ax2.set_title("Correct (Full Axis)")

plt.tight_layout()
plt.show()
```
