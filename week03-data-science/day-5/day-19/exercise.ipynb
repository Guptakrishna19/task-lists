## 1. Streamlit Widgets

Widgets allow users to interact with the application. Streamlit variables are updated dynamically when a user interacts with a widget, triggering a re-run of the script from top to bottom.

````python
# Exercise 1: Demonstrating basic widgets
import streamlit as st

# Text inputs
name = st.text_input("Enter your name", value="Alice")

# Slider
age = st.slider("Select your age", min_value=0, max_value=100, value=25)

# Selectbox
country = st.selectbox("Select your country", options=["USA", "India", "Germany", "Japan"])

# Checkbox
agree = st.checkbox("I agree to the terms")

# Multiselect
hobbies = st.multiselect("Select your hobbies", options=["Coding", "Reading", "Traveling", "Gaming"])

# Button
if st.button("Submit"):
    st.write(f"Hello {name}! You are {age} years old and live in {country}.")
    if agree:
        st.write("You agreed to the terms.")
    if hobbies:
        st.write(f"Your hobbies: {', '.join(hobbies)}")
````

## 2. Streamlit Charts

Streamlit supports native charts (like `st.line_chart`, `st.bar_chart`, `st.scatter_chart`) as well as third-party charts (Matplotlib, Seaborn, Altair, Plotly).

````python
# Exercise 2: Demonstrating charts in Streamlit
import streamlit as st
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Generate mock data
np.random.seed(42)
chart_data = pd.DataFrame(
    np.random.randn(20, 3),
    columns=['A', 'B', 'C']
)

# Streamlit native line chart
st.subheader("Streamlit Native Line Chart")
st.line_chart(chart_data)

# Matplotlib / Seaborn Chart
st.subheader("Matplotlib & Seaborn Chart")
fig, ax = plt.subplots()
sns.histplot(chart_data['A'], kde=True, ax=ax)
st.pyplot(fig)
````

## 3. Streamlit Layout

Layout elements help organize the dashboard interface. Key layout elements include columns, sidebar, tabs, and expanders.

````python
# Exercise 3: Structuring layout in Streamlit
import streamlit as st

# Sidebar layout
st.sidebar.header("Configuration Sidebar")
theme = st.sidebar.radio("Choose App Theme", options=["Light", "Dark"])

# Main layout with columns
col1, col2 = st.columns(2)
with col1:
    st.metric(label="Active Users", value="1,245", delta="+12%")
with col2:
    st.metric(label="Conversion Rate", value="2.8%", delta="-0.4%")

# Tabs layout
tab1, tab2 = st.tabs(["Tab 1", "Tab 2"])
with tab1:
    st.write("Content for Tab 1")
with tab2:
    st.write("Content for Tab 2")

# Expander
with st.expander("Show Details"):
    st.write("Here is some extra detailed information that remains hidden by default.")
````

## 4. Streamlit Caching

Caching is crucial for dashboard performance, especially when loading datasets or running heavy calculations. Streamlit provides:
- `@st.cache_data`: For caching data computations (e.g. CSV loading, SQL queries, pandas transformations).
- `@st.cache_resource`: For caching shared resources (e.g. ML models, database connections).

````python
# Exercise 4: Performance Caching in Streamlit
import streamlit as st
import time
import pandas as pd

# Caching a data loading operation
@st.cache_data
def load_large_dataset():
    # Simulate a slow network or IO operation
    time.sleep(2)
    df = pd.DataFrame({
        'Product': ['A', 'B', 'C'],
        'Sales': [100, 200, 150]
    })
    return df

st.write("Loading data...")
start_time = time.time()
data = load_large_dataset()
duration = time.time() - start_time
st.write(f"Data loaded in {duration:.4f} seconds!")
st.dataframe(data)

# Caching a resource (like a model or DB connection)
@st.cache_resource
def load_model():
    time.sleep(2)
    return "Mock ML Model"

st.write("Loading model...")
start_time = time.time()
model = load_model()
duration = time.time() - start_time
st.write(f"Model loaded in {duration:.4f} seconds!")
st.write(model)
````
