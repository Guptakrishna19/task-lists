Implement forward pass + sigmoid + MSE in NumPy; do one manual gradient step.

```python
import numpy as np

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def mse_loss(y_true, y_pred):
    return np.mean((y_true - y_pred) ** 2)

# Input data and weights
x = np.array([1.0, 2.0])
w = np.array([0.5, -0.5])
b = 0.1
y_true = 1.0

# Forward pass
z = np.dot(w, x) + b
y_pred = sigmoid(z)
loss = mse_loss(y_true, y_pred)

print(f'z: {z:.4f}')
print(f'y_pred: {y_pred:.4f}')
print(f'loss: {loss:.4f}')

# Manual gradient step
# 1. dLoss/dy_pred
dL_dy_pred = -2 * (y_true - y_pred) # Since N=1, mean is just the value, deriv of (1-y)^2 is -2*(1-y)

# 2. dy_pred/dz (sigmoid derivative)
dy_pred_dz = y_pred * (1 - y_pred)

# 3. dz/dw and dz/db
dz_dw = x
dz_db = 1

# Chain rule
dL_dw = dL_dy_pred * dy_pred_dz * dz_dw
dL_db = dL_dy_pred * dy_pred_dz * dz_db

print(f'\nGradient wrt weights (dL/dw): {dL_dw}')
print(f'Gradient wrt bias (dL/db): {dL_db}')

# Gradient descent step
learning_rate = 0.1
w_new = w - learning_rate * dL_dw
b_new = b - learning_rate * dL_db

print(f'\nUpdated weights: {w_new}')
print(f'Updated bias: {b_new}')
```

- **What was difficult:** Calculating the chain rule manually required careful tracking of dimensions.
- **What I improved:** Solidified intuition behind how each variable affects the loss.
